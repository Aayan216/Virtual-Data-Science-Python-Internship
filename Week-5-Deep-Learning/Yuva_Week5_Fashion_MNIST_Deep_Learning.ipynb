with an improved Convolutional Neural Network (CNN) for classifying

- Model comparison

```python
# ============================================================
# WEEK 5 - DEEP LEARNING APPLICATION IN DATA SCIENCE
# Fashion-MNIST Image Classification using CNN
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# Reproducibility
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 70)
print("WEEK 5 - DEEP LEARNING APPLICATION")
print("=" * 70)

print("\nTensorFlow version:", tf.__version__)

print("\nGPU devices:")
print(tf.config.list_physical_devices("GPU"))

if tf.config.list_physical_devices("GPU"):
    print("GPU acceleration is available.")
else:
    print("GPU not detected. CPU will be used.")


# ============================================================
# 2. PROJECT CONFIGURATION
# ============================================================

PROJECT_DIR = "fashion_mnist_outputs"

os.makedirs(PROJECT_DIR, exist_ok=True)

IMG_HEIGHT = 28
IMG_WIDTH = 28
NUM_CLASSES = 10

BATCH_SIZE = 128
EPOCHS = 20

print("\nProject configuration")
print("-" * 40)
print("Image size:", IMG_HEIGHT, "x", IMG_WIDTH)
print("Number of classes:", NUM_CLASSES)
print("Batch size:", BATCH_SIZE)
print("Maximum epochs:", EPOCHS)


# ============================================================
# 3. LOAD FASHION-MNIST DATASET
# ============================================================

from tensorflow.keras.datasets import fashion_mnist

print("\n" + "=" * 70)
print("LOADING FASHION-MNIST DATASET")
print("=" * 70)

(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

print("\nDataset loaded successfully!")

print("\nOriginal training data:")
print("Images:", X_train_full.shape)
print("Labels:", y_train_full.shape)

print("\nOriginal test data:")
print("Images:", X_test.shape)
print("Labels:", y_test.shape)


# ============================================================
# 4. CLASS DEFINITIONS
# ============================================================

class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]

class_mapping = pd.DataFrame({
    "Class ID": range(NUM_CLASSES),
    "Class Name": class_names
})

print("\nClass mapping:")
display(class_mapping)


# ============================================================
# 5. DATA QUALITY INSPECTION
# ============================================================

print("\n" + "=" * 70)
print("DATA QUALITY INSPECTION")
print("=" * 70)

print("\nTraining image shape:", X_train_full.shape)
print("Test image shape:", X_test.shape)

print("\nPixel data type:", X_train_full.dtype)

print("Minimum pixel value:", X_train_full.min())
print("Maximum pixel value:", X_train_full.max())

print("\nMissing values:")
print("Training images:", np.isnan(X_train_full).sum())
print("Test images:", np.isnan(X_test).sum())

print("\nLabel range:")
print("Minimum label:", y_train_full.min())
print("Maximum label:", y_train_full.max())

print("\nUnique training classes:", np.unique(y_train_full))
print("Unique test classes:", np.unique(y_test))


# ============================================================
# 6. CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION")
print("=" * 70)

train_class_counts = pd.Series(
    y_train_full
).value_counts().sort_index()

class_distribution = pd.DataFrame({
    "Class ID": range(NUM_CLASSES),
    "Class Name": class_names,
    "Training Samples": train_class_counts.values
})

display(class_distribution)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=class_distribution,
    x="Class Name",
    y="Training Samples"
)

plt.title("Fashion-MNIST Training Class Distribution")
plt.xlabel("Clothing Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "class_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 7. SAMPLE IMAGE VISUALIZATION
# ============================================================

print("\nDisplaying sample images...")

plt.figure(figsize=(12, 8))

for i in range(20):

    plt.subplot(4, 5, i + 1)

    plt.imshow(
        X_train_full[i],
        cmap="gray"
    )

    plt.title(
        class_names[y_train_full[i]]
    )

    plt.axis("off")

plt.suptitle(
    "Sample Fashion-MNIST Images",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "sample_images.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. PIXEL DISTRIBUTION
# ============================================================

sample_pixels = X_train_full[:10000].flatten()

plt.figure(figsize=(10, 6))

plt.hist(
    sample_pixels,
    bins=50
)

plt.title(
    "Distribution of Pixel Intensities"
)

plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "pixel_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 9. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

print("\n" + "=" * 70)
print("CREATING TRAIN / VALIDATION / TEST SETS")
print("=" * 70)

# 50,000 training samples
X_train = X_train_full[:50000]
y_train = y_train_full[:50000]

# 10,000 validation samples
X_val = X_train_full[50000:]
y_val = y_train_full[50000:]

print("\nTraining set:", X_train.shape)
print("Validation set:", X_val.shape)
print("Test set:", X_test.shape)


# ============================================================
# 10. DATA PREPROCESSING
# ============================================================

print("\n" + "=" * 70)
print("DATA PREPROCESSING")
print("=" * 70)

# Normalize pixels from [0, 255] to [0, 1]
X_train = X_train.astype("float32") / 255.0
X_val = X_val.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
X_train = X_train[..., np.newaxis]
X_val = X_val[..., np.newaxis]
X_test = X_test[..., np.newaxis]

print("\nAfter preprocessing:")
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

print("\nPixel range after normalization:")
print("Minimum:", X_train.min())
print("Maximum:", X_train.max())


# ============================================================
# 11. BASELINE MLP MODEL
# ============================================================

print("\n" + "=" * 70)
print("BASELINE MODEL - MULTI-LAYER PERCEPTRON")
print("=" * 70)

baseline_model = keras.Sequential([

    layers.Input(
        shape=(28, 28, 1)
    ),

    layers.Flatten(),

    layers.Dense(
        256,
        activation="relu"
    ),

    layers.Dropout(0.20),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.20),

    layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )

], name="Baseline_MLP")

baseline_model.summary()


# ============================================================
# 12. COMPILE BASELINE
# ============================================================

baseline_model.compile(

    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)

baseline_callbacks = [

    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),

    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]


# ============================================================
# 13. TRAIN BASELINE
# ============================================================

print("\nTraining baseline MLP...")

baseline_history = baseline_model.fit(

    X_train,
    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    callbacks=baseline_callbacks,

    verbose=1
)


# ============================================================
# 14. TRAINING HISTORY FUNCTION
# ============================================================

def plot_training_history(
    history,
    model_name
):

    history_df = pd.DataFrame(
        history.history
    )

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14, 5)
    )

    # Accuracy
    axes[0].plot(
        history_df["accuracy"],
        label="Training Accuracy"
    )

    axes[0].plot(
        history_df["val_accuracy"],
        label="Validation Accuracy"
    )

    axes[0].set_title(
        f"{model_name} - Accuracy"
    )

    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Accuracy")

    axes[0].legend()

    # Loss
    axes[1].plot(
        history_df["loss"],
        label="Training Loss"
    )

    axes[1].plot(
        history_df["val_loss"],
        label="Validation Loss"
    )

    axes[1].set_title(
        f"{model_name} - Loss"
    )

    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Loss")

    axes[1].legend()

    plt.tight_layout()

    filename = (
        model_name
        .lower()
        .replace(" ", "_")
        + "_learning_curves.png"
    )

    plt.savefig(
        os.path.join(
            PROJECT_DIR,
            filename
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 15. BASELINE LEARNING CURVES
# ============================================================

plot_training_history(
    baseline_history,
    "Baseline MLP"
)


# ============================================================
# 16. MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model,
    X_data,
    y_data,
    model_name
):

    probabilities = model.predict(
        X_data,
        batch_size=BATCH_SIZE,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    accuracy = accuracy_score(
        y_data,
        predictions
    )

    precision = precision_score(
        y_data,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_data,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_data,
        predictions,
        average="weighted",
        zero_division=0
    )

    print(
        f"\n{model_name}"
    )

    print("=" * 50)

    print(
        f"Accuracy : {accuracy:.4f}"
    )

    print(
        f"Precision: {precision:.4f}"
    )

    print(
        f"Recall   : {recall:.4f}"
    )

    print(
        f"F1 Score : {f1:.4f}"
    )

    return {

        "Model": model_name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1

    }, predictions, probabilities


# ============================================================
# 17. EVALUATE BASELINE
# ============================================================

baseline_results, baseline_predictions, baseline_probabilities = evaluate_model(

    baseline_model,

    X_test,

    y_test,

    "Baseline MLP"

)


# ============================================================
# 18. IMPROVED CNN MODEL
# ============================================================

print("\n" + "=" * 70)
print("IMPROVED CNN MODEL")
print("=" * 70)

data_augmentation = keras.Sequential([

    layers.RandomRotation(
        0.05
    ),

    layers.RandomTranslation(
        0.05,
        0.05
    ),

    layers.RandomZoom(
        0.05
    )

], name="Data_Augmentation")


cnn_model = keras.Sequential([

    layers.Input(
        shape=(28, 28, 1)
    ),

    # Data augmentation
    data_augmentation,

    # Convolution Block 1
    layers.Conv2D(
        32,
        kernel_size=(3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation(
        "relu"
    ),

    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    layers.Dropout(
        0.20
    ),

    # Convolution Block 2
    layers.Conv2D(
        64,
        kernel_size=(3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation(
        "relu"
    ),

    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    layers.Dropout(
        0.25
    ),

    # Convolution Block 3
    layers.Conv2D(
        128,
        kernel_size=(3, 3),
        padding="same"
    ),

    layers.BatchNormalization(),

    layers.Activation(
        "relu"
    ),

    # Global feature aggregation
    layers.GlobalAveragePooling2D(),

    # Classification head
    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(
        0.40
    ),

    layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )

], name="Improved_CNN")

cnn_model.summary()


# ============================================================
# 19. CNN ARCHITECTURE DIAGRAM
# ============================================================

try:

    keras.utils.plot_model(

        cnn_model,

        to_file=os.path.join(
            PROJECT_DIR,
            "cnn_architecture.png"
        ),

        show_shapes=True,

        show_layer_names=True,

        expand_nested=True,

        dpi=120

    )

    print(
        "\nCNN architecture diagram saved successfully."
    )

    from IPython.display import Image, display

    display(

        Image(

            filename=os.path.join(
                PROJECT_DIR,
                "cnn_architecture.png"
            )

        )

    )

except Exception as e:

    print(
        "\nArchitecture diagram could not be generated."
    )

    print(
        "Model summary is available above."
    )

    print(
        "Reason:",
        e
    )


# ============================================================
# 20. COMPILE CNN
# ============================================================

cnn_model.compile(

    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)

cnn_callbacks = [

    keras.callbacks.EarlyStopping(

        monitor="val_loss",

        patience=4,

        restore_best_weights=True,

        verbose=1

    ),

    keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.5,

        patience=2,

        min_lr=1e-6,

        verbose=1

    )

]


# ============================================================
# 21. TRAIN CNN
# ============================================================

print("\nTraining improved CNN...")

cnn_history = cnn_model.fit(

    X_train,

    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    callbacks=cnn_callbacks,

    verbose=1

)


# ============================================================
# 22. CNN LEARNING CURVES
# ============================================================

plot_training_history(
    cnn_history,
    "Improved CNN"
)


# ============================================================
# 23. EVALUATE CNN
# ============================================================

cnn_results, cnn_predictions, cnn_probabilities = evaluate_model(

    cnn_model,

    X_test,

    y_test,

    "Improved CNN"

)


# ============================================================
# 24. MODEL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

results_df = pd.DataFrame([

    baseline_results,

    cnn_results

])

display(results_df)

results_df.to_csv(

    os.path.join(
        PROJECT_DIR,
        "model_comparison.csv"
    ),

    index=False

)


# ============================================================
# 25. DETAILED CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("CNN CLASSIFICATION REPORT")
print("=" * 70)

report = classification_report(

    y_test,

    cnn_predictions,

    target_names=class_names,

    digits=4

)

print(report)

report_dict = classification_report(

    y_test,

    cnn_predictions,

    target_names=class_names,

    output_dict=True

)

classification_report_df = pd.DataFrame(
    report_dict
).transpose()

classification_report_df.to_csv(

    os.path.join(
        PROJECT_DIR,
        "cnn_classification_report.csv"
    )

)

display(
    classification_report_df
)


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    cnn_predictions

)

plt.figure(
    figsize=(12, 9)
)

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names

)

plt.title(
    "Improved CNN - Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    rotation=45
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(

    os.path.join(
        PROJECT_DIR,
        "cnn_confusion_matrix.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 27. PER-CLASS ACCURACY
# ============================================================

per_class_accuracy = []

for class_id in range(NUM_CLASSES):

    mask = (
        y_test == class_id
    )

    correct = np.sum(

        cnn_predictions[mask]
        ==
        y_test[mask]

    )

    total = np.sum(mask)

    accuracy = (
        correct / total
    )

    per_class_accuracy.append({

        "Class ID": class_id,

        "Class": class_names[class_id],

        "Correct": correct,

        "Total": total,

        "Accuracy": accuracy

    })

per_class_df = pd.DataFrame(
    per_class_accuracy
)

per_class_df["Accuracy (%)"] = (
    per_class_df["Accuracy"] * 100
)

display(
    per_class_df
)


# ============================================================
# 28. PER-CLASS ACCURACY VISUALIZATION
# ============================================================

plt.figure(
    figsize=(12, 6)
)

sns.barplot(

    data=per_class_df,

    x="Class",

    y="Accuracy (%)"

)

plt.title(
    "CNN Accuracy by Clothing Class"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.xticks(
    rotation=45
)

plt.ylim(
    0,
    100
)

plt.tight_layout()

plt.savefig(

    os.path.join(
        PROJECT_DIR,
        "per_class_accuracy.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 29. MOST CONFUSED CLASS PAIRS
# ============================================================

confusion_pairs = []

for actual in range(NUM_CLASSES):

    for predicted in range(NUM_CLASSES):

        if actual != predicted:

            count = cm[
                actual,
                predicted
            ]

            confusion_pairs.append({

                "Actual":
                    class_names[actual],

                "Predicted":
                    class_names[predicted],

                "Errors":
                    count

            })

confusion_pairs_df = pd.DataFrame(
    confusion_pairs
)

confusion_pairs_df = (
    confusion_pairs_df
    .sort_values(
        by="Errors",
        ascending=False
    )
)

print(
    "\nMost frequent misclassification pairs:"
)

display(
    confusion_pairs_df.head(10)
)


# ============================================================
# 30. ERROR ANALYSIS
# ============================================================

incorrect_indices = np.where(

    cnn_predictions
    !=
    y_test

)[0]

print(
    "\nTotal incorrect predictions:",
    len(incorrect_indices)
)

plt.figure(
    figsize=(14, 10)
)

num_examples = min(
    20,
    len(incorrect_indices)
)

for i in range(num_examples):

    index = incorrect_indices[i]

    plt.subplot(
        4,
        5,
        i + 1
    )

    plt.imshow(

        X_test[index].squeeze(),

        cmap="gray"

    )

    actual_label = (
        class_names[
            y_test[index]
        ]
    )

    predicted_label = (
        class_names[
            cnn_predictions[index]
        ]
    )

    confidence = (

        np.max(
            cnn_probabilities[index]
        )
        * 100

    )

    plt.title(

        f"Actual: {actual_label}\n"
        f"Pred: {predicted_label}\n"
        f"Conf: {confidence:.1f}%"

    )

    plt.axis(
        "off"
    )

plt.suptitle(

    "Examples of Incorrect CNN Predictions",

    fontsize=16

)

plt.tight_layout()

plt.savefig(

    os.path.join(
        PROJECT_DIR,
        "cnn_error_examples.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 31. PREDICTION CONFIDENCE ANALYSIS
# ============================================================

prediction_confidence = np.max(

    cnn_probabilities,

    axis=1

)

confidence_df = pd.DataFrame({

    "Actual": [

        class_names[i]

        for i in y_test

    ],

    "Predicted": [

        class_names[i]

        for i in cnn_predictions

    ],

    "Confidence":
        prediction_confidence,

    "Correct":
        (
            cnn_predictions
            ==
            y_test
        )

})

print(
    "\nAverage prediction confidence:"
)

print(

    f"{confidence_df['Confidence'].mean() * 100:.2f}%"

)

print(
    "\nAverage confidence for correct predictions:"
)

print(

    f"{confidence_df.loc[confidence_df['Correct'], 'Confidence'].mean() * 100:.2f}%"

)

print(
    "\nAverage confidence for incorrect predictions:"
)

print(

    f"{confidence_df.loc[~confidence_df['Correct'], 'Confidence'].mean() * 100:.2f}%"

)


# ============================================================
# 32. OVERFITTING ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("OVERFITTING ANALYSIS")
print("=" * 70)

history_df = pd.DataFrame(
    cnn_history.history
)

best_epoch = (
    history_df["val_accuracy"].idxmax()
    + 1
)

best_val_accuracy = (
    history_df["val_accuracy"].max()
)

final_train_accuracy = (
    history_df["accuracy"].iloc[-1]
)

final_val_accuracy = (
    history_df["val_accuracy"].iloc[-1]
)

accuracy_gap = (

    final_train_accuracy
    -
    final_val_accuracy

)

print(
    "Best epoch:",
    best_epoch
)

print(

    f"Best validation accuracy: "
    f"{best_val_accuracy:.4f}"

)

print(

    f"Final training accuracy: "
    f"{final_train_accuracy:.4f}"

)

print(

    f"Final validation accuracy: "
    f"{final_val_accuracy:.4f}"

)

print(

    f"Training-validation accuracy gap: "
    f"{accuracy_gap:.4f}"

)

if accuracy_gap > 0.05:

    print(

        "\nInterpretation: A noticeable gap exists "
        "between training and validation accuracy, "
        "indicating possible overfitting."

    )

else:

    print(

        "\nInterpretation: Training and validation "
        "performance remain relatively close, "
        "suggesting that overfitting is reasonably controlled."

    )


# ============================================================
# 33. TRAIN / VALIDATION / TEST COMPARISON
# ============================================================

test_accuracy = (
    cnn_results["Accuracy"]
)

comparison = pd.DataFrame({

    "Dataset": [

        "Training",

        "Validation",

        "Test"

    ],

    "Accuracy": [

        final_train_accuracy,

        final_val_accuracy,

        test_accuracy

    ]

})

display(
    comparison
)

plt.figure(
    figsize=(8, 5)
)

sns.barplot(

    data=comparison,

    x="Dataset",

    y="Accuracy"

)

plt.title(
    "CNN Accuracy Across Dataset Splits"
)

plt.ylabel(
    "Accuracy"
)

plt.ylim(
    0,
    1
)

plt.tight_layout()

plt.savefig(

    os.path.join(
        PROJECT_DIR,
        "train_validation_test_accuracy.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 34. SAVE TRAINING HISTORY
# ============================================================

history_df.to_csv(

    os.path.join(
        PROJECT_DIR,
        "cnn_training_history.csv"
    ),

    index=False

)


# ============================================================
# 35. SAVE TEST PREDICTIONS
# ============================================================

prediction_output = pd.DataFrame({

    "Actual_Label_ID":
        y_test,

    "Actual_Class": [

        class_names[i]

        for i in y_test

    ],

    "Predicted_Label_ID":
        cnn_predictions,

    "Predicted_Class": [

        class_names[i]

        for i in cnn_predictions

    ],

    "Confidence":
        prediction_confidence,

    "Correct": (

        cnn_predictions
        ==
        y_test

    )

})

prediction_output.to_csv(

    os.path.join(
        PROJECT_DIR,
        "fashion_mnist_predictions.csv"
    ),

    index=False

)

print(
    "\nPrediction file saved."
)

display(
    prediction_output.head(10)
)


# ============================================================
# 36. SAVE MODEL METRICS
# ============================================================

final_metrics = pd.DataFrame([

    baseline_results,

    cnn_results

])

final_metrics.to_csv(

    os.path.join(
        PROJECT_DIR,
        "fashion_mnist_model_metrics.csv"
    ),

    index=False

)

display(
    final_metrics
)


# ============================================================
# 37. SAVE CNN MODEL
# ============================================================

model_path = os.path.join(

    PROJECT_DIR,

    "fashion_mnist_cnn.keras"

)

cnn_model.save(
    model_path
)

print(
    "\nModel saved to:",
    model_path
)


# ============================================================
# 38. CNN WITHOUT DATA AUGMENTATION
# ============================================================

print("\n" + "=" * 70)
print("EXPERIMENT - CNN WITHOUT DATA AUGMENTATION")
print("=" * 70)

cnn_no_aug = keras.Sequential([

    layers.Input(
        shape=(28, 28, 1)
    ),

    layers.Conv2D(
        32,
        (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.MaxPooling2D(
        (2, 2)
    ),

    layers.Dropout(
        0.20
    ),

    layers.Conv2D(
        64,
        (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.MaxPooling2D(
        (2, 2)
    ),

    layers.Dropout(
        0.25
    ),

    layers.Conv2D(
        128,
        (3, 3),
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(
        0.40
    ),

    layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )

], name="CNN_No_Augmentation")


cnn_no_aug.compile(

    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)

no_aug_callbacks = [

    keras.callbacks.EarlyStopping(

        monitor="val_loss",

        patience=3,

        restore_best_weights=True

    ),

    keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.5,

        patience=2,

        min_lr=1e-6

    )

]

no_aug_history = cnn_no_aug.fit(

    X_train,

    y_train,

    validation_data=(

        X_val,

        y_val

    ),

    epochs=15,

    batch_size=BATCH_SIZE,

    callbacks=no_aug_callbacks,

    verbose=1

)


# ============================================================
# 39. EVALUATE CNN WITHOUT AUGMENTATION
# ============================================================

no_aug_results, no_aug_predictions, no_aug_probabilities = evaluate_model(

    cnn_no_aug,

    X_test,

    y_test,

    "CNN Without Augmentation"

)


# ============================================================
# 40. FINAL EXPERIMENT COMPARISON
# ============================================================

experiment_results = pd.DataFrame([

    baseline_results,

    no_aug_results,

    cnn_results

])

experiment_results = (

    experiment_results

    .sort_values(

        by="F1 Score",

        ascending=False

    )

)

print("\n" + "=" * 70)
print("FINAL EXPERIMENT COMPARISON")
print("=" * 70)

display(
    experiment_results
)

experiment_results.to_csv(

    os.path.join(
        PROJECT_DIR,
        "fashion_mnist_experiment_results.csv"
    ),

    index=False

)


# ============================================================
# 41. EXPERIMENT VISUALIZATION
# ============================================================

plot_df = experiment_results.melt(

    id_vars="Model",

    value_vars=[

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score"

    ],

    var_name="Metric",

    value_name="Score"

)

plt.figure(
    figsize=(12, 6)
)

sns.barplot(

    data=plot_df,

    x="Model",

    y="Score",

    hue="Metric"

)

plt.title(
    "Comparison of Deep Learning Models"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=20
)

plt.tight_layout()

plt.savefig(

    os.path.join(
        PROJECT_DIR,
        "model_experiment_comparison.png"
    ),

    dpi=300,

    bbox_inches="tight"

)

plt.show()


# ============================================================
# 42. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL WEEK 5 PROJECT SUMMARY")
print("=" * 70)

print("\nDataset:")
print("Fashion-MNIST")

print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Validation samples:",
    len(X_val)
)

print(
    "Test samples:",
    len(X_test)
)

print(
    "\nNumber of classes:",
    NUM_CLASSES
)

print("\nFinal CNN Test Performance")
print("-" * 40)

print(

    f"Accuracy : "
    f"{cnn_results['Accuracy']:.4f}"

)

print(

    f"Precision: "
    f"{cnn_results['Precision']:.4f}"

)

print(

    f"Recall   : "
    f"{cnn_results['Recall']:.4f}"

)

print(

    f"F1 Score : "
    f"{cnn_results['F1 Score']:.4f}"

)

print(

    "\nBest Validation Accuracy:",
    f"{best_val_accuracy:.4f}"

)

print(

    "Total Test Errors:",
    np.sum(
        cnn_predictions != y_test
    )

)

print(

    "\nOutput directory:",
    PROJECT_DIR

)

print("\n" + "=" * 70)
print("WEEK 5 NOTEBOOK EXECUTION COMPLETE")
print("=" * 70)

print(
    "\nAll generated files are inside:",
    PROJECT_DIR
)
```

```python
# ============================================================
# WEEK 5 - LIGHT DATA AUGMENTATION EXPERIMENT
# ============================================================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))

# ------------------------------------------------------------
# 1. LIGHT DATA AUGMENTATION
# ------------------------------------------------------------

light_augmentation = keras.Sequential([
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.05)
], name="Light_Data_Augmentation")


# ------------------------------------------------------------
# 2. BUILD LIGHT-AUGMENTATION CNN
# ------------------------------------------------------------

light_cnn = keras.Sequential([

    layers.Input(shape=(28, 28, 1)),

    # Light augmentation
    light_augmentation,

    # Block 1
    layers.Conv2D(32, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.15),

    # Block 2
    layers.Conv2D(64, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.ReLU(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.20),

    # Block 3
    layers.Conv2D(128, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.ReLU(),

    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.25),

    layers.Dense(10, activation="softmax")

], name="Light_Augmentation_CNN")


# ------------------------------------------------------------
# 3. COMPILE
# ------------------------------------------------------------

light_cnn.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 4. CALLBACKS
# ------------------------------------------------------------

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=1,
        min_lr=1e-6
    )
]


# ------------------------------------------------------------
# 5. TRAIN
# ------------------------------------------------------------

print("\nTraining Light-Augmentation CNN...")

history_light = light_cnn.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=5,
    batch_size=128,
    callbacks=callbacks,
    verbose=1
)


# ------------------------------------------------------------
# 6. TEST EVALUATION
# ------------------------------------------------------------

test_loss, test_accuracy = light_cnn.evaluate(
    X_test,
    y_test,
    verbose=0
)

probabilities = light_cnn.predict(
    X_test,
    batch_size=256,
    verbose=0
)

predictions = np.argmax(probabilities, axis=1)

precision = precision_score(
    y_test,
    predictions,
    average="weighted"
)

recall = recall_score(
    y_test,
    predictions,
    average="weighted"
)

f1 = f1_score(
    y_test,
    predictions,
    average="weighted"
)


# ------------------------------------------------------------
# 7. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("LIGHT-AUGMENTATION CNN RESULTS")
print("=" * 55)

print(f"Test Accuracy : {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Precision     : {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall        : {recall:.4f} ({recall*100:.2f}%)")
print(f"F1 Score      : {f1:.4f} ({f1*100:.2f}%)")


# ------------------------------------------------------------
# 8. COMPARE ALL THREE MODELS
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Model": [
        "Baseline MLP",
        "CNN without Augmentation",
        "CNN with Heavy Augmentation",
        "CNN with Light Augmentation"
    ],
    "Accuracy": [
        0.8929,
        0.9191,
        0.8703,
        test_accuracy
    ],
    "Precision": [
        0.8936,
        0.9200,
        0.8823,
        precision
    ],
    "Recall": [
        0.8929,
        0.9191,
        0.8703,
        recall
    ],
    "F1 Score": [
        0.8931,
        0.9193,
        0.8730,
        f1
    ]
})

print("\nMODEL COMPARISON")
display(comparison)


# ------------------------------------------------------------
# 9. SAVE RESULTS
# ------------------------------------------------------------

os.makedirs("fashion_mnist_outputs", exist_ok=True)

comparison.to_csv(
    "fashion_mnist_outputs/final_model_comparison.csv",
    index=False
)

pd.DataFrame({
    "Epoch": range(1, len(history_light.history["accuracy"]) + 1),
    "Training Accuracy": history_light.history["accuracy"],
    "Validation Accuracy": history_light.history["val_accuracy"],
    "Training Loss": history_light.history["loss"],
    "Validation Loss": history_light.history["val_loss"]
}).to_csv(
    "fashion_mnist_outputs/light_augmentation_history.csv",
    index=False
)


# ------------------------------------------------------------
# 10. PLOT COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

x = np.arange(len(comparison))
width = 0.2

plt.bar(
    x - 1.5 * width,
    comparison["Accuracy"],
    width,
    label="Accuracy"
)

plt.bar(
    x - 0.5 * width,
    comparison["Precision"],
    width,
    label="Precision"
)

plt.bar(
    x + 0.5 * width,
    comparison["Recall"],
    width,
    label="Recall"
)

plt.bar(
    x + 1.5 * width,
    comparison["F1 Score"],
    width,
    label="F1 Score"
)

plt.xticks(
    x,
    comparison["Model"],
    rotation=15,
    ha="right"
)

plt.ylabel("Score")
plt.ylim(0.80, 1.00)
plt.title("Final Comparison of Deep Learning Experiments")
plt.legend()
plt.tight_layout()

plt.savefig(
    "fashion_mnist_outputs/final_model_comparison.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 11. LEARNING CURVES
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    history_light.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history_light.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Light-Augmentation CNN - Accuracy")
plt.legend()


plt.subplot(1, 2, 2)

plt.plot(
    history_light.history["loss"],
    label="Training Loss"
)

plt.plot(
    history_light.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Light-Augmentation CNN - Loss")
plt.legend()

plt.tight_layout()

plt.savefig(
    "fashion_mnist_outputs/light_augmentation_learning_curves.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 12. FINAL MESSAGE
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("EXPERIMENT COMPLETED")
print("=" * 55)

print("\nSaved:")
print("✓ final_model_comparison.csv")
print("✓ final_model_comparison.png")
print("✓ light_augmentation_history.csv")
print("✓ light_augmentation_learning_curves.png")

print("\nDo NOT write the report yet.")
print("Send me the results/table from this cell first.")
```

```python
# ============================================================
# FINAL RESULTS & REPORT EXPORT CELL
# IMPORTANT:
# - DOES NOT TRAIN ANY MODEL
# - DOES NOT CALL model.fit()
# - DOES NOT RETRAIN
# - ONLY USES MODELS/RESULTS ALREADY IN MEMORY
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from tensorflow import keras

# ------------------------------------------------------------
# 1. CREATE OUTPUT DIRECTORY
# ------------------------------------------------------------

OUTPUT_DIR = "fashion_mnist_final_results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 70)
print("FASHION-MNIST FINAL RESULTS EXPORT")
print("=" * 70)
print("No training will be performed.")
print("No model.fit() will be called.")
print()


# ------------------------------------------------------------
# 2. FIND EXISTING DATASET VARIABLES
# ------------------------------------------------------------

required_data = ["X_test", "y_test"]

missing_data = [x for x in required_data if x not in globals()]

if missing_data:
    raise RuntimeError(
        f"Required variables not found: {missing_data}\n"
        "Make sure the earlier dataset cells have already been executed."
    )

print("✓ Test dataset found")
print(f"  X_test shape: {X_test.shape}")
print(f"  y_test shape: {y_test.shape}")
print()


# ------------------------------------------------------------
# 3. AUTOMATICALLY FIND EXISTING KERAS MODELS
# ------------------------------------------------------------

models_found = {}

for name, obj in list(globals().items()):

    if isinstance(obj, keras.Model):

        # Ignore temporary/internal models
        if not name.startswith("_"):
            models_found[name] = obj


if len(models_found) == 0:
    raise RuntimeError(
        "No trained Keras models were found in memory.\n"
        "Do NOT retrain yet. Check that the model-training cell "
        "was executed successfully before this cell."
    )


print("✓ Existing Keras models detected:")
for name in models_found:
    print(f"  • {name}")

print()


# ------------------------------------------------------------
# 4. DETERMINE WHICH MODELS ARE USABLE
# ------------------------------------------------------------

# Prefer meaningful project models if they exist
preferred_order = [
    "baseline_model",
    "cnn_model",
    "improved_cnn",
    "final_model",
    "model"
]

ordered_models = {}

for name in preferred_order:
    if name in models_found:
        ordered_models[name] = models_found[name]

# Add any remaining models
for name, model in models_found.items():
    if name not in ordered_models:
        ordered_models[name] = model


# ------------------------------------------------------------
# 5. EVALUATE EXISTING MODELS
# ------------------------------------------------------------

results = []
predictions_store = {}

print("=" * 70)
print("EVALUATING EXISTING MODELS")
print("=" * 70)

for model_name, model in ordered_models.items():

    print(f"\nEvaluating: {model_name}")

    try:

        # IMPORTANT:
        # evaluate() and predict() do NOT train the model.
        test_result = model.evaluate(
            X_test,
            y_test,
            verbose=0
        )

        # Get predictions
        probabilities = model.predict(
            X_test,
            verbose=0
        )

        predictions = np.argmax(
            probabilities,
            axis=1
        )

        accuracy = accuracy_score(
            y_test,
            predictions
        )

        precision = precision_score(
            y_test,
            predictions,
            average="weighted",
            zero_division=0
        )

        recall = recall_score(
            y_test,
            predictions,
            average="weighted",
            zero_division=0
        )

        f1 = f1_score(
            y_test,
            predictions,
            average="weighted",
            zero_division=0
        )

        results.append({
            "Model": model_name,
            "Test Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1 Score": f1,
            "Keras Test Loss": float(test_result[0])
        })

        predictions_store[model_name] = {
            "probabilities": probabilities,
            "predictions": predictions
        }

        print(f"  Accuracy : {accuracy:.4f}")
        print(f"  Precision: {precision:.4f}")
        print(f"  Recall   : {recall:.4f}")
        print(f"  F1 Score : {f1:.4f}")
        print(f"  Loss     : {float(test_result[0]):.4f}")

    except Exception as e:

        print(f"  ⚠ Could not evaluate {model_name}")
        print(f"  Reason: {e}")


# ------------------------------------------------------------
# 6. CREATE RESULTS TABLE
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

if results_df.empty:
    raise RuntimeError(
        "Models were found, but none could be evaluated."
    )

print()
print("=" * 70)
print("MODEL RESULTS")
print("=" * 70)

display(
    results_df.style.format({
        "Test Accuracy": "{:.4%}",
        "Precision": "{:.4%}",
        "Recall": "{:.4%}",
        "F1 Score": "{:.4%}",
        "Keras Test Loss": "{:.4f}"
    })
)


# Save CSV
results_df.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False
)


# ------------------------------------------------------------
# 7. SAVE CLASSIFICATION REPORTS
# ------------------------------------------------------------

class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]

for model_name, data in predictions_store.items():

    predictions = data["predictions"]

    report = classification_report(
        y_test,
        predictions,
        target_names=class_names,
        digits=4,
        zero_division=0
    )

    safe_name = model_name.replace("/", "_")

    with open(
        f"{OUTPUT_DIR}/{safe_name}_classification_report.txt",
        "w"
    ) as f:
        f.write(report)

    print(f"\n{'=' * 70}")
    print(f"{model_name} - Classification Report")
    print("=" * 70)
    print(report)


# ------------------------------------------------------------
# 8. CONFUSION MATRIX FOR BEST EXISTING MODEL
# ------------------------------------------------------------

best_model_name = results_df.loc[
    results_df["Test Accuracy"].idxmax(),
    "Model"
]

best_predictions = predictions_store[
    best_model_name
]["predictions"]

cm = confusion_matrix(
    y_test,
    best_predictions
)

plt.figure(figsize=(11, 9))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    f"Confusion Matrix - {best_model_name}"
)

plt.colorbar()

plt.xticks(
    range(10),
    class_names,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(10),
    class_names
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/final_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 9. PER-CLASS ACCURACY
# ------------------------------------------------------------

per_class_accuracy = (
    np.diag(cm) /
    cm.sum(axis=1)
) * 100

class_accuracy_df = pd.DataFrame({
    "Class": class_names,
    "Accuracy (%)": per_class_accuracy
})

print()
print("=" * 70)
print("PER-CLASS ACCURACY")
print("=" * 70)

display(
    class_accuracy_df.style.format({
        "Accuracy (%)": "{:.2f}"
    })
)

class_accuracy_df.to_csv(
    f"{OUTPUT_DIR}/per_class_accuracy.csv",
    index=False
)


# ------------------------------------------------------------
# 10. PER-CLASS ACCURACY GRAPH
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.bar(
    class_names,
    per_class_accuracy
)

plt.title(
    f"Per-Class Accuracy - {best_model_name}"
)

plt.xlabel("Clothing Class")
plt.ylabel("Accuracy (%)")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.ylim(0, 100)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/per_class_accuracy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 11. FIND MOST CONFUSED CLASS PAIRS
# ------------------------------------------------------------

confusion_pairs = []

for i in range(10):

    for j in range(10):

        if i != j:

            confusion_pairs.append({
                "Actual": class_names[i],
                "Predicted": class_names[j],
                "Count": cm[i, j]
            })

confusion_pairs_df = pd.DataFrame(
    confusion_pairs
).sort_values(
    "Count",
    ascending=False
)

print()
print("=" * 70)
print("TOP 10 CONFUSION PAIRS")
print("=" * 70)

display(
    confusion_pairs_df.head(10)
)

confusion_pairs_df.to_csv(
    f"{OUTPUT_DIR}/confusion_pairs.csv",
    index=False
)


# ------------------------------------------------------------
# 12. SAVE BEST MODEL PREDICTIONS
# ------------------------------------------------------------

best_probabilities = predictions_store[
    best_model_name
]["probabilities"]

best_predictions = predictions_store[
    best_model_name
]["predictions"]

prediction_df = pd.DataFrame({
    "Actual_Label": y_test,
    "Predicted_Label": best_predictions,
    "Actual_Class": [
        class_names[x] for x in y_test
    ],
    "Predicted_Class": [
        class_names[x] for x in best_predictions
    ],
    "Confidence": np.max(
        best_probabilities,
        axis=1
    )
})

prediction_df.to_csv(
    f"{OUTPUT_DIR}/final_predictions.csv",
    index=False
)


# ------------------------------------------------------------
# 13. ERROR ANALYSIS SUMMARY
# ------------------------------------------------------------

incorrect = prediction_df[
    prediction_df["Actual_Label"] !=
    prediction_df["Predicted_Label"]
]

correct = prediction_df[
    prediction_df["Actual_Label"] ==
    prediction_df["Predicted_Label"]
]

print()
print("=" * 70)
print("ERROR ANALYSIS")
print("=" * 70)

print(f"Correct predictions : {len(correct)}")
print(f"Incorrect predictions: {len(incorrect)}")
print(
    f"Error rate           : "
    f"{len(incorrect) / len(prediction_df) * 100:.2f}%"
)

print()
print("Most common errors:")

display(
    incorrect.groupby(
        ["Actual_Class", "Predicted_Class"]
    )
    .size()
    .reset_index(name="Count")
    .sort_values(
        "Count",
        ascending=False
    )
    .head(10)
)


# ------------------------------------------------------------
# 14. CHECK MODEL ARCHITECTURE
# ------------------------------------------------------------

print()
print("=" * 70)
print(f"BEST MODEL ARCHITECTURE: {best_model_name}")
print("=" * 70)

best_model = ordered_models[best_model_name]

best_model.summary()


# Save architecture JSON
with open(
    f"{OUTPUT_DIR}/best_model_architecture.json",
    "w"
) as f:

    f.write(
        best_model.to_json()
    )


# ------------------------------------------------------------
# 15. CHECK FOR EXISTING TRAINING HISTORY
# ------------------------------------------------------------

history_candidates = {}

for name, obj in list(globals().items()):

    if isinstance(obj, keras.callbacks.History):

        history_candidates[name] = obj


print()
print("=" * 70)
print("EXISTING TRAINING HISTORIES")
print("=" * 70)

if history_candidates:

    for name in history_candidates:

        print(f"✓ {name}")

        history = history_candidates[name]

        history_df = pd.DataFrame(
            history.history
        )

        history_df.to_csv(
            f"{OUTPUT_DIR}/{name}_history.csv",
            index=False
        )

else:

    print(
        "No History object found in memory."
    )

    print(
        "This is okay — existing model evaluation "
        "and analysis are still saved."
    )


# ------------------------------------------------------------
# 16. CREATE REPORT-READY SUMMARY
# ------------------------------------------------------------

best_row = results_df.loc[
    results_df["Test Accuracy"].idxmax()
]

summary = {
    "Dataset": "Fashion-MNIST",
    "Number of Classes": 10,
    "Test Samples": int(len(y_test)),
    "Best Existing Model": best_model_name,
    "Test Accuracy": float(
        best_row["Test Accuracy"]
    ),
    "Precision": float(
        best_row["Precision"]
    ),
    "Recall": float(
        best_row["Recall"]
    ),
    "F1 Score": float(
        best_row["F1 Score"]
    ),
    "Test Loss": float(
        best_row["Keras Test Loss"]
    ),
    "Correct Predictions": int(
        len(correct)
    ),
    "Incorrect Predictions": int(
        len(incorrect)
    )
}

with open(
    f"{OUTPUT_DIR}/final_summary.json",
    "w"
) as f:

    json.dump(
        summary,
        f,
        indent=4
    )


# ------------------------------------------------------------
# 17. FINAL OUTPUT
# ------------------------------------------------------------

print()
print("=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print(f"Best model       : {best_model_name}")
print(
    f"Test accuracy    : "
    f"{summary['Test Accuracy']:.4%}"
)
print(
    f"Precision        : "
    f"{summary['Precision']:.4%}"
)
print(
    f"Recall           : "
    f"{summary['Recall']:.4%}"
)
print(
    f"F1 Score         : "
    f"{summary['F1 Score']:.4%}"
)
print(
    f"Correct samples  : "
    f"{summary['Correct Predictions']}"
)
print(
    f"Incorrect samples: "
    f"{summary['Incorrect Predictions']}"
)

print()
print("=" * 70)
print("FILES SAVED")
print("=" * 70)

for filename in sorted(
    os.listdir(OUTPUT_DIR)
):

    print(
        f"✓ {filename}"
    )

print()
print(
    f"All final results are inside: "
    f"{OUTPUT_DIR}/"
)

print()
print("IMPORTANT: No model was trained in this cell.")
```
